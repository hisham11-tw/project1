## 21. Export Species Summary
Calculate species-level mean values and save them to a CSV file.

```python
species_summary = df.groupby("species")[["sepal_length", "sepal_width", "petal_length", "petal_width"]].mean()
```

```python
species_summary.to_csv("output/species_summary.csv")
```

## 22. Export Filtered Iris Data

```python
filtered_iris = df[df["sepal_length"] > 6]
```
